```python
GAME_PROFILE = "pocket"  # pocket | tcg
```

```python
from pathlib import Path
import sys

config_by_profile = {"pocket": "pocket.yaml", "tcg": "tcg.yaml"}
if GAME_PROFILE not in config_by_profile:
    raise ValueError("GAME_PROFILE must be 'pocket' or 'tcg'")

cwd = Path.cwd().resolve()
BASE_DIR = next(
    (candidate for candidate in (cwd, cwd.parent)
     if (candidate / "config" / config_by_profile[GAME_PROFILE]).is_file()
     and (candidate / "pipelines" / "deck_ranking.py").is_file()),
    None,
)
if BASE_DIR is None:
    raise RuntimeError("Open this notebook from the repository root or its notebooks directory")
if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

from pipelines.deck_ranking import run_deck_ranking

CONFIG_PATH = BASE_DIR / "config" / config_by_profile[GAME_PROFILE]
```

```python
result = run_deck_ranking(
    base_dir=BASE_DIR,
    config_path=CONFIG_PATH,
    run_scrape=True,
    run_core=True,
    run_mars=True,
    run_heatmap=True,
    run_report=True,
    show_progress=True,
)
result
```
